## 共用資料準備
以下在記憶體重建與既有教材相同的資料。執行後可以直接開始本單元；不會覆寫 CSV。


````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

````python
# 重現兩項可核對的證據：品質與平均成績
print("形狀：", df.shape)
print("欄位缺失：\n", df.isna().sum())
print("平均成績：", df["quiz_score"].mean())
print("睡眠與成績有效配對：", len(df.dropna(subset=["sleep_hours", "quiz_score"])))
````

````python
# 學生小任務 MB：請在此撰寫你的程式
# 執行前先確認上方資料準備與必要cell已執行；可新增其他程式cell。
# TODO：依任務要求加入程式；完成後執行並保留輸出。
# -------------------------------------------------
# 0️⃣  前置：載入常用套件
# -------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 設定圖表樣式（可依個人喜好調整）
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# -------------------------------------------------
# 1️⃣  載入資料（請將下行換成你實際的讀取方式）
# -------------------------------------------------
# 例如：df = pd.read_csv("my_data.csv", parse_dates=["timestamp"])
# 假設你已經有一個名為 df 的 DataFrame，且包含以下欄位：
#   - subject_id   : 受試者/樣本編號（用於分組）
#   - timestamp    : 記錄時間（必須為 datetime64[ns] 類型）
#   - value        : 需要計算平均的數值變數
#   - treatment    : 兩種處理方式的標籤（例如 "A"、"B"）
#   - pair_key     : 用於配對的鍵（可為空，之後會自行產生）
# 若欄位名稱不同，請自行更改下面程式中的對應名稱。

# 以下僅示意：如果你還沒有 df，請先用下面這行產出一個極簡的範例資料框（僅供練習，實際分析請用真實資料）
# df = pd.DataFrame({
#     "subject_id": np.repeat([1,2,3], 5),
#     "timestamp": pd.date_range("2026-01-01", periods=15, freq="H"),
#     "value": np.random.rand(15),
#     "treatment": np.random.choice(["A","B"], size=15),
# })

# -------------------------------------------------
# 2️⃣  基本檢查與清理
# -------------------------------------------------
print("=== 資料概覽 ===")
print(df.info())                     # 欄位型別、非 null 數量
print("\n前五筆資料：")
print(df.head())                     # 先目視確認欄位內容

# 檢查缺失值
missing = df.isnull().sum()
print("\n缺失值統計：")
print(missing[missing > 0])          # 只有有缺失的欄位會顯示

# （依需求）處理缺失值——這裡僅示範刪除缺失 timestamp 或 value 的列
# 你可以改為填補、標記等等，請依實際情況決定。
df = df.dropna(subset=["timestamp", "value"])

# 確認 timestamp 已經是 datetime 類型；若不是，則轉換
if not pd.api.types.is_datetime64_any_dtype(df["timestamp"]):
    df["timestamp"] = pd.to_datetime(df["timestamp"])

# -------------------------------------------------
# 3️⃣  分組：計算每組的人數與平均值
# -------------------------------------------------
# 假設我們要依照 subject_id 與 treatment 兩個欄位分組
group_cols = ["subject_id", "treatment"]
grouped = (
    df.groupby(group_cols)
      .agg(
          count=("value", "size"),      # 組內觀測數（人數或記錄數）
          mean_value=("value", "mean")  # 組內 value 的平均
      )
      .reset_index()
)

print("\n=== 分組人數與平均 ===")
print(grouped)

# 檢查方式：觀察 count 是否與你預期的每組記錄數相符；
#          mean_value 是否在合理範圍內（例如不會是極端的 NaN 或無限大）。

# -------------------------------------------------
# 4️⃣  有效配對：在 15 小時窗內找出同一 subject_id 不同 treatment 的配對
# -------------------------------------------------
# 這裡示範一種「最近鄰」配對概念：對每個 subject_id，
#   找出 timestamp 差距 ≤ 15 小時且 treatment 不同的記錄對。
#   若你有其他配對規則（例如依據特定 ID、實驗批次等），請自行調整。

def find_pairs_within_window(sub_df, window_hours=15):
    """
    給定單一 subject_id 的子資料框，回傳所有符合時間窗且 treatment 不同的配對。
    每列代表一個配對 (idx1, idx2, time_diff)。
    """
    pairs = []
    # 依時間排序，方便後續滑動窗口
    sub_df = sub_df.sort_values("timestamp").reset_index(drop=True)
    n = len(sub_df)
    for i in range(n):
        ti = sub_df.loc[i, "timestamp"]
        ti_treat = sub_df.loc[i, "treatment"]
        # 只看往後的記錄，避免重複計算 (i,j) 與 (j,i)
        for j in range(i+1, n):
            tj = sub_df.loc[j, "timestamp"]
            if (tj - ti) > pd.Timedelta(hours=window_hours):
                # 超過窗口，後面的只會更大，直接跳出內層迴圈
                break
            if sub_df.loc[j, "treatment"] != ti_treat:
                pairs.append({
                    "idx1": sub_df.index[i],
                    "idx2": sub_df.index[j],
                    "time_diff_h": (tj - ti).total_seconds() / 3600,
                    "treatment_i": ti_treat,
                    "treatment_j": sub_df.loc[j, "treatment"],
                    "value_i": sub_df.loc[i, "value"],
                    "value_j": sub_df.loc[j, "value"]
                })
    return pd.DataFrame(pairs)

# 對每個 subject_id 分別跑配對函式
pair_list = []
for sid, sub_df in df.groupby("subject_id"):
    pairs_df = find_pairs_within_window(sub_df, window_hours=15)
    if not pairs_df.empty:
        pairs_df["subject_id"] = sid
        pair_list.append(pairs_df)

if pair_list:
    pairs_all = pd.concat(pair_list, ignore_index=True)
    print("\n=== 有效配對（≤15 小時、不同 treatment） ===")
    print(pairs_all.head())          # 只顯示前幾列，以免刷屏
    print(f"總共找到 {len(pairs_all)} 筆配對")
else:
    print("\n=== 在 15 小時窗內沒有找到任何符合條件的配對 ===")
    pairs_all = pd.DataFrame()       # 空的 DataFrame，後續仍可安全使用

# 檢查方式：
#   - 確認 time_diff_h 欄位皆 ≤ 15（若有超出，表示邏輯錯誤）
#   - treatment_i 與 treatment_j 必須不同
#   - 若你有預期的配對數目（例如實驗設計說明），可與此數字比較。

# -------------------------------------------------
# 5️⃣  兩種處理方式的 15 小時紀錄比較
# -------------------------------------------------
# 目的：分別看 treatment = A 與 treatment = B 在最近 15 小時的 sliding window 內，
#       value 的平均值如何隨時間變化，進而畫出兩條曲線作比較。

def rolling_mean_15h(sub_df, value_col="value", time_col="timestamp"):
    """
    給定已排序的子資料框，計算 sliding window (15 小時) 內的平均值。
    回傳一個 Series，索引為原始時間點，值為該點以前 15 小時的平均。
    """
    sub_df = sub_df.sort_values(time_col).set_index(time_col)
    rolling = sub_df[value_col].rolling(f"15H").mean()
    return rolling

# 分別對兩種 treatment 計算 rolling mean
rolling_A = rolling_mean_15h(df[df["treatment"] == "A"])
rolling_B = rolling_mean_15h(df[df["treatment"] == "B"])

# 合併成一個 DataFrame 方便繪圖
compare_df = pd.DataFrame({
    "treatment_A": rolling_A,
    "treatment_B": rolling_B
}).dropna()   # 去除窗口不足 15 小時的開頭 NaN

print("\n=== 15 小時滑動平均（前五筆） ===")
print(compare_df.head())

# 檢查方式：
#   - 確認索引為 timestamp，且單調遞增。
#   - 值應在 value 的合理範圍內，且不會出現極端的 NaN（除非資料本身太稀疏）。

# -------------------------------------------------
# 6️⃣  繪圖：長條圖（分組平均） + 折線圖（15 小時滑動平均）
# -------------------------------------------------
plt.figure(figsize=(12, 5))

# ----- 6.1 長條圖：各組的平均 value -----
plt.subplot(1, 2, 1)
sns.barplot(
    data=grouped,
    x="subject_id",
    y="mean_value",
    hue="treatment",
    ci=None   # 不顯示誤差條，僅顯示平均值
)
plt.title("各組（subject_id）平均 value")
plt.ylabel("平均 value")
plt.xlabel("subject_id")
plt.legend(title="處理方式")

# ----- 6.2 折線圖：15 小時滑動平均比較 -----
plt.subplot(1, 2, 2)
plt.plot(compare_df.index, compare_df["treatment_A"], label="處理 A", color="#1f77b4")
plt.plot(compare_df.index, compare_df["treatment_B"], label="處理 B", color="#ff7f0e")
plt.title("15 小時滑動平均 value（處理 A vs B）")
plt.ylabel("滑動平均 value")
plt.xlabel("時間")
plt.legend()
plt.tight_layout()
plt.show()

# 檢查圖表：
#   - 長條圖：每個 subject_id 下應該有兩個長條（A、B），高度代表該組的平均。
#   - 折線圖：兩條線應該隨時間平滑變化；若出現劇烈跳躍，可能代表資料稀疏或窗口計算問題。

# -------------------------------------------------
# 7️⃣  結果說明與因果推論限制提醒
# -------------------------------------------------
markdown_text = """
### 📌 結果摘要（請根據上面印出的表格與圖表填寫）

1. **分組人數與平均**
   - 每個 `subject_id` 在各 `treatment` 下的觀測數（count）與 `value` 的平均（mean_value）已在 `grouped` 表格中給出。
   - 請檢查 `count` 是否符合你實際的每組樣本數；`mean_value` 的大小應落在原始 `value` 的範圍內。

2. **有效配對（≤15 小時、不同 treatment）**
   - `pairs_all` 表格列出所有符合時間窗且 treatment 不同的配對對。
   - 欄位說明：
     - `idx1`, `idx2`：原始 `df` 的列索引
     - `subject_id`：配對所屬的受試者
     - `time_diff_h`：兩筆記錄之時間差（小時）
     - `treatment_i`, `treatment_j`：配對兩端的處理方式（必定不同）
     - `value_i`, `value_j`：對應的原始數值

3. **15 小時滑動平均比較**
   - `compare_df` 提供了每個時間點（索引）下，過去 15 小時內 `value` 的滑動平均，分別對應處理 A 與處理 B。
   - 折線圖直觀顯示兩條處理在時間上的趨勢差異。

### ⚠️ 因果推論的限制
- **此資料僅為模擬/練習用**，並未隨機分配處理或控制混雜變數。
- 觀察到的平均差異或滑動平均趨勢 **不等同於處理造成的效果**，可能受到以下因素影響：
  1. **選擇性偏 bias**（例如較健康的受試者較可能被分配到某個處理）
  2. **時間混雜**（例如隨時間的季節性、設備校準）
  3. **未測量的共變數**（例如生活習慣、基因背景）
- 因此，**僅能說明「關聯」**，而不能斷言「處理 A 導致 value 上升/下降」。若想談因果，必須隨機對照實驗（RCT）或使用適當的準實驗工具變數、縱向模型、邊界迴歸等方法，並在分析前先說明假設與檢驗方式。

> **請在報告中清楚註明：**
> “基於此觀察資料的描述性統計，我們發現 X 與 Y 之間存在關聯，但由於研究設計非隨機分配，**無法由此推論因果關係**。”
"""

# 印出 Markdown 說明（可直接複製到報告中）
from IPython.display import Markdown
display(Markdown(markdown_text))
````
